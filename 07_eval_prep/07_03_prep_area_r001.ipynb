# Preparation array with grid cell areas at res. 0.01°

```python
# Library
import xarray as xr

# Import functions to calculate grid cell area and grid cell boundaries
from xgrid_utils import grid_cell_areas
```

```python
# Directory
dir01 = '../paper_ecosystem_carbon/output/01_data_prep/'
dir07 = '../paper_ecosystem_carbon/output/07_eval_prep/'
```

```python
# Read land sea mask
da = xr.open_zarr(f'{dir01}ds_prep_copernicus_land_mask_r001.zarr') \
    .copernicus_land_mask

# Create area array, convert from m2 to ha and rechunk
da_area = xr.DataArray(data=grid_cell_areas(da.lon.data, da.lat.data),
                       dims=('lat', 'lon'), 
                       coords=(da.lat.data, da.lon.data,)) \
    .chunk(dict(lat=5000, lon=5000)) * 0.0001

da_area \
    .to_dataset(name='area_ha') \
    .to_zarr(f'{dir07}ds_prep_area_r001.zarr', mode='w');
```
