```python
# Import function to post-process blue output
from blue_postprocessing_v01 import prep_ds_eluc_components, prep_ds_eluc_pft, prep_ds_cveg_csoil
```

```python
# Directories
dir07 = '../paper_ecosystem_carbon/output/07_eval_prep/'
dir_blue_runs = '../paper_ecosystem_carbon/blue_runs/'
```

---

```python
# Define blue runs
dir_blue_run = f'{dir_blue_runs}20260515_run/'

run_default = 'run_20260515_default_1700-2024'
run_cspatial = 'run_20260515_cspatial_1700-2024'
run_cdgvm = 'run_20260515_cdgvm_1700-2024'

l_runs = [run_default, run_cspatial, run_cdgvm]

# Define component fluxes to process
components = ['net', 'sources', 'sinks', 'deforestation', 'forestation', 
              'wood_harvest_sources', 'wood_harvest_sinks', 'other_transitions']

# Create directory of processed blue files
dir07_bluep = f'{dir07}blue_processed/'
os.makedirs(dir07_bluep, exist_ok=True)
```

---

### Post-process blue data

```python
%%time
for run_blue in l_runs:
    print(run_blue)
    prep_ds_eluc_components(dir_blue_run, run_blue, components, dir07_bluep, f'{run_blue}_eluc_components')
    prep_ds_eluc_pft(dir_blue_run, run_blue, dir07_bluep, f'{run_blue}_eluc_pft')
    prep_ds_cveg_csoil(dir_blue_run, run_blue, dir07_bluep, f'{run_blue}_cveg_csoil')
```
