# Plot maps underlying carbon

```python
# Libraries
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Directories
dir02 = '../paper_ecosystem_carbon/output/02_database_prep/'
dir08 = '../paper_ecosystem_carbon/output/08_eval/'
dir_nearth = '../data/naturalearth/'
```

---

### Pre-processing

```python
# Get carbon data
ds_cveg_biom = xr.open_zarr(f'{dir02}ds_cveg_biom_r001.zarr')
ds_csoil = xr.open_zarr(f'{dir02}ds_csoil_000_100_r001.zarr')

# Get coastline data
coastline110 = gpd.read_file(
    f'{dir_nearth}ne_110m_coastline/ne_110m_coastline.shp')
```

### Plot

```python
# Plot
fig, axs = plt.subplots(figsize=(10, 5), ncols=1, nrows=2, dpi=300)
axs = axs.ravel()

# Plot cveg
ds_cveg_biom.cveg_biom.plot.imshow(ax=axs[0], cmap='Greens',
                                   vmin=0, vmax=150,
                                   cbar_kwargs=dict(label='tC ha$^{-1}$'))
axs[0].set_title('Vegetation carbon')

# Plot csoil
ds_csoil.csoil_000_100.plot.imshow(ax=axs[1], cmap='Oranges', 
                                   vmin=0, vmax=400,
                                   cbar_kwargs=dict(label='tC ha$^{-1}$'))
axs[1].set_title('Soil organic carbon')

# Layout and add coastlines
for ax in axs:    
    ax.set_ylim([-60, 90])
    coastline110.plot(ax=ax, color='#000000', linewidth=0.5)
    ax.axis('off')

# Export
fstr_out = 'fig_s10_carbon_underlying'
plt.savefig(f'{dir08}pdf/{fstr_out}.pdf', dpi=300, 
            bbox_inches='tight', pad_inches=0.1); 
plt.savefig(f'{dir08}png/{fstr_out}.png', dpi=300, 
            bbox_inches='tight', pad_inches=0.1);
```
