# Plot map IPCC regions

```python
# Libraries
import numpy as np
import pandas as pd
import xarray as xr
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
```

```python
# Directories
dir08 = '../paper_ecosystem_carbon/output/08_eval/'
dir_nearth = '../data/naturalearth/'
dir_iturbide = '../data/iturbide2020/'
```

---

### Pre-processing

```python
# Coastline and land shapefiles from natural earth
coastline110 = gpd.read_file(f'{dir_nearth}ne_110m_coastline/ne_110m_coastline.shp')
```

```python
# Get ipcc region shapefile data
geodf_ipcc = gpd.read_file(f'{dir_iturbide}'
                           'IPCC-WGI-reference-regions-v4_shapefile/'
                           'IPCC-WGI-reference-regions-v4.shp')

# Remove water regions and and antarctica
geodf_ipcc = geodf_ipcc[(geodf_ipcc['Type'].isin(['Land', 'Land-Ocean'])) & 
                        (~geodf_ipcc['Acronym'].isin(['EAN', 'WAN']))]

# List with ipcc region acronyms
list_acr = list(geodf_ipcc['Acronym'])
```

---

### Plot

```python
# Plot IPCC regions
fig, ax = plt.subplots(figsize=(15, 8))

coastline110.plot(ax=ax, color='#000000', linewidth=0.5)   
geodf_ipcc.plot(ax=ax, color='#ffffff', edgecolor='#000000')

for index, row in geodf_ipcc.iterrows():
    xy = row['geometry'].centroid.coords[:]
    xytext = row['geometry'].centroid.coords[:]
    ax.text(xy[0][0], xy[0][1], row['Acronym'], 
                fontsize=10, ha='center', va='center', weight='bold',
                bbox=dict(boxstyle='square,pad=0.05', fc='w', ec='none'))

ax.set_ylim(-60, 90)
ax.axis('off')

# Export
fstr_out = 'fig_s08_ipcc_regions'
plt.savefig(f'{dir08}pdf/{fstr_out}.pdf', dpi=300, 
            bbox_inches='tight', pad_inches=0.1); 
plt.savefig(f'{dir08}png/{fstr_out}.png', dpi=300, 
            bbox_inches='tight', pad_inches=0.1);
```
