# Regridding DGVM files

```python
# Libraries
import os, subprocess, tempfile, gzip, tarfile
import xarray as xr
```

```python
# Directories
dir_trendy = '../TRENDY/trendyv12-gcb2023/'
dir04 = '../paper_ecosystem_carbon/output/04_dgvm_prep/'
```

---

```python
def regridd_dgvm_to_r025(model, input_dir, output_dir, input_file, output_file):

    """Regridds TRENDY V12 data to LUH2 resolution with cdo remapycon"""

    # Target file
    target_file = '../data/luh2_v2h/states.nc'
    
    # Ensure the output directory exists
    os.makedirs(output_dir, exist_ok=True)

    # Create a temporary bash script file in the output directory
    with tempfile.NamedTemporaryFile(delete=False, mode='w', suffix='.sh', 
                                     dir=output_dir) as temp_script:
        temp_script_path = temp_script.name
        temp_script.write(
            f"""
            #!/bin/bash
            module load cdo
            cdo -P 24 remapycon,"{target_file}" "{input_dir}{input_file}" "{output_dir}{output_file}"
        """)

    try:
        # Make the script executable
        os.chmod(temp_script_path, 0o755)

        # Run the script
        subprocess.run(["bash", temp_script_path], check=True)

    finally:
        # Delete the script after execution
        os.remove(temp_script_path)
```

```python
# Cable-pop
model = 'CABLEPOP'
input_dir = f'{dir_trendy}{model}/S2/'
output_dir = f'{dir04}trendy_v12_r025/{model}/'

# Ensure the output directory exists
os.makedirs(output_dir, exist_ok=True)

for i in ['cSoilpft', 'cVegpft', 'landCoverFrac']:
    input_file = f'CABLE-POP_S2_{i}.nc'
    output_file = f'{model}_S2_{i}_r025.nc'

    # Get zipped nc file and export as nc file
    with gzip.open(f'{input_dir}{input_file}.gz') as fp:
        xr.open_dataset(fp, decode_times=False) \
            .to_netcdf(f'{output_dir}{input_file}', mode='w')
    
    # Regridding (note input_dir is output_dir, so unzipped nc file is used)
    regridd_dgvm_to_r025(model, output_dir, output_dir, input_file, output_file)
    
    # Remove unzipped nc file 
    os.remove(f'{output_dir}{input_file}')
```

```python
# Yibs
model = 'YIBS'
input_dir = f'{dir_trendy}{model}/S2/'
output_dir = f'{dir04}trendy_v12_r025/{model}/'


for i in ['cSoilpft', 'cVegpft', 'landCoverFrac']:
    input_file = f'YIBs_S2_Annual_{i}.nc'
    output_file = f'{model}_S2_{i}_r025.nc'

    # Extract tar file to output dir
    with tarfile.open(f'{input_dir}{input_file}.tar.gz', "r:gz") as tar:
        tar.extractall(path=output_dir)

    # Regridding (note input_dir is output_dir, so unzipped nc file is used)
    regridd_dgvm_to_r025(model, output_dir, output_dir, input_file, output_file)
    
    # Remove extracted nc file 
    os.remove(f'{output_dir}{input_file}')
```
